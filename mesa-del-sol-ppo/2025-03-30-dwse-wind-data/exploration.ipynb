```python
import os
import pandas as pd

def list_csv_columns(root_dir):
    # Recursively find all CSV files
    for dirpath, _, filenames in os.walk(root_dir):
        for file in filenames:
            if file.endswith(".csv"):
                file_path = os.path.join(dirpath, file)
                try:
                    df = pd.read_csv(file_path, nrows=1)  # Read only the header
                    print(f"{file_path}:")
                    print(", ".join(df.columns))
                    print("-" * 50)
                except Exception as e:
                    print(f"Error reading {file_path}: {e}")
root_directory = "."
list_csv_columns(root_directory)
```

### Wind Spatial Datasets: Contains wind data for 10 turbines. Rows represent time points, columns are Turbine #1-10 with values showing wind measurements.
### Wind Time Series Datasets: Hourly/10-minute data with Time, WindSpeed, and Power columns. Shows wind speed and corresponding power output over time for a single turbine.
### Inland Wind Farm Datasets: 10-minute data from turbines WT1-WT4, collected 2010-2011. Contains Sequence, wind speed (V), direction (D), air density, turbulence (I), below-hub shear (S_b), and power output (y).
### Offshore Wind Farm Datasets: Similar to Inland, but includes humidity (H) and above-hub shear (S_a). From turbines WT5-WT6, collected in 2009.
### Wind Spatio-Temporal Datasets: Contains average and standard deviation of wind speeds from 120 turbines (V1-V120) for 2009-2010. Rows are hourly timestamps, columns represent different turbines.
### Wind Spatio-Temporal Datasets: Extended dataset with 200 turbines plus 3 measurement masts.

```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv('./Wind_Spatio_Temporal_Dataset1/Wind Spatio-Temporal Dataset1(2009, Ave).csv')
columns_to_plot = ['V1', 'V2', 'V3', 'V4', 'V5', 'V6', 'V7', 'V8', 'V9', 'V10']
plt.figure(figsize=(10, 6))

for column in columns_to_plot:
    plt.plot(data.index, data[column], label=column)

# Labeling the axes and adding title
plt.xlabel('Time')
plt.ylabel('Wind Speed (V)')
plt.title('Wind Speed for Various Turbines (2009 Average)')

# Show legend
plt.legend()

# Show the plot
plt.grid(True)
plt.show()
```
