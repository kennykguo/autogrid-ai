# Summary

The UCSD microgrid supports a 450-hectare campus with 45,000 daily occupants, supplying 85% of electricity, 95% of heating, and 95% of cooling. 

Generation: Two 13.5 MW gas turbines, one 3 MW steam turbine, a 1.2 MW solar array, and a 2.8 MW molten carbonate fuel cell using waste methane

HVAC & Storage: A 140,674 kWh, 14,385 m³ thermal energy storage bank, with three steam-driven chillers and five electric chillers

Grid & Control: Connected to SDG&E via a 69 kV substation with a SCADA system for energy coordination, Paladin master controller VPower

```python
import os
import pandas as pd

def list_csv_columns(root_dir):
    # Recursively find all CSV files
    for dirpath, _, filenames in os.walk(root_dir):
        for file in filenames:
            if file.endswith(".csv"):
                file_path = os.path.join(dirpath, file)
                try:
                    df = pd.read_csv(file_path, nrows=1)  # Read only the header
                    print(f"{file_path}:")
                    print(", ".join(df.columns))
                    print("-" * 50)
                except Exception as e:
                    print(f"Error reading {file_path}: {e}")

# Change this to your actual directory
root_directory = "."
list_csv_columns(root_directory)
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt

# Enable inline plotting in Jupyter Notebook
%matplotlib inline  

def list_csv_columns(base_dir):
    """Recursively lists all CSV files and their column names."""
    csv_info = {}
    
    for root, _, files in os.walk(base_dir):
        for file in files:
            if file.endswith(".csv"):
                filepath = os.path.join(root, file)
                try:
                    df = pd.read_csv(filepath, nrows=1)  # Read only first row for column names
                    csv_info[filepath] = list(df.columns)
                except Exception as e:
                    print(f"Skipping {filepath}: {e}")
    
    return csv_info

def plot_csv(filepath):
    """Reads a CSV file, assumes a time-based index, and plots all variables inline."""
    try:
        df = pd.read_csv(filepath)
        
        # Identify possible time column
        time_col = None
        for col in ["DateTime", "Time", "time"]:
            if col in df.columns:
                time_col = col
                break
        
        if not time_col:
            print(f"Skipping {filepath}: No recognized time column found.")
            return
        
        df[time_col] = pd.to_datetime(df[time_col], errors='coerce')  # Convert to datetime
        df = df.dropna(subset=[time_col])  # Drop rows with invalid timestamps
        df.set_index(time_col, inplace=True)
        
        # Ensure at least one numeric column exists
        numeric_cols = df.select_dtypes(include=['number']).columns
        if len(numeric_cols) == 0:
            print(f"Skipping {filepath}: No numeric columns to plot.")
            return
        
        for column in numeric_cols:
            plt.figure(figsize=(10, 5))
            plt.plot(df.index, df[column], label=column)
            plt.xlabel('Time')
            plt.ylabel(column)
            plt.title(f"{os.path.basename(filepath)} - {column}")
            plt.legend()
            plt.grid()
            plt.show()  # Show the plot inline in Jupyter Notebook
            
    except Exception as e:
        print(f"Skipping {filepath}: {e}")

def process_directory(base_dir):
    """Recursively processes directories and plots CSV files inline."""
    for root, _, files in os.walk(base_dir):
        for file in files:
            if file.endswith(".csv"):
                filepath = os.path.join(root, file)
                plot_csv(filepath)

# Define main directory
base_directory = "."  # Update this to your main directory

# List all CSV columns
csv_columns = list_csv_columns(base_directory)

# Print structured column information
for file, columns in csv_columns.items():
    print(f"{file}:\n  {', '.join(columns)}\n{'-'*50}")

# Process and plot data inline
process_directory(base_directory)
```
